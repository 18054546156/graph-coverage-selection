# MedMNIST-C reliability results

This notebook reads completed `metrics.csv` files. It does not select data, train models, or generate corruptions.

```python
from pathlib import Path
import pandas as pd

RESULTS = Path('../runs')
files = sorted(RESULTS.glob('**/metrics.csv'))
summary = []
for path in files:
    frame = pd.read_csv(path)
    clean = frame[frame['corruption'].eq('clean')]
    corrupt = frame[frame['corruption'].ne('clean')]
    if not clean.empty and not corrupt.empty:
        summary.append({'run': str(path), 'clean_ba': clean['ba'].iloc[0], 'corruption_mean_ba': corrupt['ba'].mean()})
pd.DataFrame(summary)
```
